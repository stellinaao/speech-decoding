```python
import numpy as np
import pickle
import matplotlib.pyplot as plt

def get_test_stats(model_name, verbose=False):
    fpath = '../logs/speech_logs/' + model_name + '/trainingStats'
    with open(fpath, 'rb') as f:
        stats = pickle.load(f)
    if verbose: 
        print(f"{model_name}: ")
        print(f"loss: {stats['testLoss'][-1]}")
        print(f"per: {stats['testCER'][-1]}")
    return stats

def plot(stats, title):
    
    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(5, 3))
    axes[0].plot(stats['testLoss'])
    axes[0].plot(np.diff(stats['testLoss']))
    axes[0].set_title(np.round(stats['testLoss'][-1], 4))

    axes[1].plot(stats['testCER'])
    axes[1].plot(np.diff(stats['testCER']))
    axes[1].set_title(np.round(stats['testCER'][-1], 4))

    fig.suptitle(title)

    plt.tight_layout()
    plt.show()
    
    return f"Batch of Lowest CER: {np.argmin(stats['testCER']) * 100}"
```

# GRU

```python
stats_gru = get_test_stats('gru')
plot(stats_gru, "GRU")
```

# LSTMs 

```python
stats_lstm_five = get_test_stats('lstm_five')
plot(stats_lstm_five, "LSTM")
```

```python
stats_ae_gru = get_test_stats('ae_gru')
plot(stats_ae_gru, title="AE, GRU")
```
